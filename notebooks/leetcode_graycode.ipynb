```python
import pandas as pd

def graycode(order):
    table = pd.DataFrame({
        "position": range(len(order)),
        "decimal": order,
    })

    table["binary"] = table["decimal"].map(
        lambda value: format(value, f"0{n}b")
    )

    table["next_binary"] = (
        table["binary"]
        .shift(-1, fill_value=table.loc[0, "binary"])
    )

    table["bits_changed_to_next"] = [
        sum(left != right for left, right in zip(current, following))
        for current, following
        in zip(table["binary"], table["next_binary"])
    ]

    return table
```

```python
n = 2
orders = [[0, 1, 2, 3], [0, 1, 3, 2]]

for order in orders: 
    table = graycode(order)
    print(order, table['bits_changed_to_next'].eq(1).all())
```

```python
order = [3, 2, 1]

def answer(order):
    table = pd.DataFrame({
        "position": range(1, len(order) + 1),
        "value": order,
    })

    table["value_divisible_by_position"] = (
        table["value"] % table["position"] == 0
    )

    table["position_divisible_by_value"] = (
        table["position"] % table["value"] == 0
    )

    table["valid"] = (
        table["value_divisible_by_position"]
        | table["position_divisible_by_value"]
    )

    return table
```

```python
n = 2
orders = [[1, 2, 3], [1, 3, 2]]

for order in orders: 
    table = answer(order)
    print(order, table['valid'].eq(1).all())
```
