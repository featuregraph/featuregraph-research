```python
import numpy as np
import matplotlib.pyplot as plt

V = np.array([[1,1], [-2,2], [4,-7]])
origin = np.array([[0, 0, 0],[0, 0, 0]]) # origin point

plt.quiver(*origin, V[:,0], V[:,1], color=['r','b','g'], scale=21)
```

```python
# plot the span of a vector

v = np.array(((0, 0), (-2, 2), (0, 0)))
# get a vector u that is orthogonal to v
u = np.array(((2, 2), (0, 0), (0, 0)))
# plot the span of u
plt.axline((0,0), slope=2, color='r')
origin = np.array(((0, 0, 0),(0, 0, 0)))
plt.quiver(*origin, v[:, 0], v[:, 1], color='b')
plt.quiver(*origin, u[:, 0], u[:, 1], color='b')

# show that this red line is a subspace
# do this exercise in 2 dimensions
```
