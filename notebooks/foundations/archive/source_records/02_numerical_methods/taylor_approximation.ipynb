```python
import torch

def taylor_approximation(f, x, a=0.0, degree=3):
    # Ensure a tensor
    a_tensor = torch.tensor(a, dtype=x.dtype, device=x.device, requires_grad=True)
    
    # Initialize approximation with 0
    approximation = torch.zeros_like(x)
    
    # Compute Taylor series
    for n in range(degree + 1):
        # Evaluate function at a
        y = f(a_tensor)
        
        # Compute n-th derivative
        for _ in range(n):
            grads = torch.autograd.grad(y, a_tensor, create_graph=True)[0]
            y = grads
        
        # Add n-th term: f^n(a)/n! * (x - a)^n
        term = y / torch.factorial(torch.tensor(n, dtype=x.dtype)) * (x - a) ** n
        approximation = approximation + term
    
    return approximation

f = lambda x: torch.sin(x)
x = torch.linspace(-1, 1, 100)
y_approx = taylor_approximation(f, x, a=0.0, degree=5)

import matplotlib.pyplot as plt
plt.plot(x.numpy(), torch.sin(x).numpy(), label="sin(x)")
plt.plot(x.numpy(), y_approx.detach().numpy(), label="Taylor Approx")
plt.legend()
plt.show()
```
