```python
import numpy as np
import matplotlib.pyplot as plt
import math
import pandas as pd
```

```python
n = 100
y_shift = -100
x_shift = 0

zero = np.zeros(n)
one = np.ones(n)
rect_width = (2*math.pi/n)
x = np.linspace(-math.pi, math.pi, n, endpoint=False)
y = np.sin(x + x_shift) + y_shift

A = y - y_shift
mask = A > 0

print((A.dot(one)*rect_width).round(3))
print((A[mask].dot(one[mask])*rect_width).round(3))

plt.figure()
plt.plot(x, y)
```
