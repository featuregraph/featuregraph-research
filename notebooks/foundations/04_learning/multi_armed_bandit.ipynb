```python
# solve mab problem using only linear transformations
```

```python
Q = np.zeros((5))
N = np.zeros((5))
R = [-1, 1, -2, 2, 0]
```

```python
Q
```

```python
for i in range(5):
    print(i)
    epsilon = random.random()
    if epsilon < 0.5:
        A = np.random.choice(np.arange(Q.size)) # pick a random arm if all Q values are equal
    else:
        A = np.argmax(Q) 
    reward = R[i]
    N[A] += 1
    Q[A] += (1/N[A])*(reward - Q[A])
```
